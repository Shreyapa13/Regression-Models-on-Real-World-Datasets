```python
#X = np.array(df["Date"])
X = np.array(range(1, len(df)+1), dtype=float)

Y = np.array(df["Close"])
```

```python
class LinearRegression:

    def __init__(self):
        self.m = 0
        self.c = 0

    def fit(self, X, Y):

        n = len(X)

        sum_x = np.sum(X)
        sum_y = np.sum(Y)

        sum_xy = np.sum(X * Y)
        sum_x2 = np.sum(X ** 2)

        self.m = ((n * sum_xy) - (sum_x * sum_y)) / \
                 ((n * sum_x2) - (sum_x ** 2))

        self.c = (sum_y - self.m * sum_x) / n

    def predict(self, X):

        return self.m * X + self.c
```

```python
model = LinearRegression()

model.fit(X, Y)

print("Slope =", model.m)
print("Intercept =", model.c)
```

```python
trendline = model.predict(X)
```

```python
plt.figure(figsize=(10,5))

plt.scatter(X, Y, label="Actual Price")

plt.plot(X, trendline, label="Trendline")

plt.xlabel("Day")
plt.ylabel("Close Price")

plt.title("NVDA Trend Analysis")

plt.legend()

plt.grid(True)

plt.show()
```

```python
corr = np.corrcoef(X, Y)[0,1]

print("Correlation =", round(corr,4))
```

```python
ss_total = np.sum((Y - np.mean(Y))**2)

ss_residual = np.sum((Y - trendline)**2)

r2 = 1 - (ss_residual / ss_total)

print("R² =", round(r2,4))
```

```python
print("\n----- REPORT -----")

print("Slope:", round(model.m,4))
print("Correlation:", round(corr,4))
print("R²:", round(r2,4))

if model.m > 0:
    print("Trend: UPWARD")
else:
    print("Trend: DOWNWARD")
```
